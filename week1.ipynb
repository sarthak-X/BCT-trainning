```python
import numpy as np
```

```python
x=np.array([1,2,3,4,5],dtype=float)
y=np.array([20,40,60,80,100],dtype=float)
w=5.0
b=0.0
learning_rate=1
for epoch in range(1000):

#prediction
 prediction=w*x+b
#loss
 error=y-prediction
 loss=np.mean(error**2)
#gradinets
 dw=-2*np.mean(x*error)
 db=-2*np.mean(error)

#update
 w-=learning_rate*dw
 b-=learning_rate*db
 if epoch%100==0:
   new_loss=np.mean((y-(w*x+b))**2)
   print(
      epoch,
      "loss=",new_loss,
      "w=", w,
      "b=",b
  )
```

First Backpropagation Implementation

```python
import numpy as np
```

```python
x=2.0
y=10.0
w=3.0
b=1.0
learning_rate=0.1

#forword pass
z=w*x+b
error=z-y
loss=error**2
print("prediction:",z)
print("error:",error)
print("loss:",loss)
 #backword pass
dL_dyhat=-2*error
dyhat_dw=x
dyhat_db=1
dw=dL_dyhat*dyhat_dw
db=dL_dyhat*dyhat_db
print("dw:",dw)
print("db:",db)

#learning rate
w-=learning_rate*dw
b-=learning_rate*db

print("new_w:",w)
print("new_b:",b)

```
